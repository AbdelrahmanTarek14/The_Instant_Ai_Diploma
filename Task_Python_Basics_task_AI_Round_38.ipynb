(14) Write a Python program that takes two numbers and an operator (+, -, x, /, //) from the user, then performs the selected operation and displays the result.**
```
Example:

Input first number: 12
Input second number: 5
Input operator (+, -, *, /, //): +

Output:
Result = 17
```

````python
a=int(input('frist number: '))
b=int(input('second number: '))
c=input('operator (+, -, *, /, //):')
print('Output:')
if c == '+':
  print('Result =',a+b)
elif c == '-' :
  print('Result =',a-b)
elif c =='*' :
  print('Result =',a*b)
elif c == '/' :
  print('Result =',a/b)
elif c == '//' :
  print('Result =',a//b)
else:
  print('invaild')
````
